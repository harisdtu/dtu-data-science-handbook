```python
# @title Load and process temperature-energy data
import pandas as pd
import os
import urllib.request
import matplotlib.pyplot as plt
import numpy as np
!pip install xicorpy

# Download from Google Drive (direct link format)
file_id = "1hAxfF6C5Sw-xSzQEgt86xcyn7VRUtvrx"
url = f"https://drive.google.com/uc?export=download&id={file_id}"

urllib.request.urlretrieve(url, 'temperature_energy_data.csv')

# Load dataset
file_P = os.path.join(os.getcwd(), 'temperature_energy_data.csv')
df = pd.read_csv(file_P)

print(df.head())
print(df.columns)
```

```python
# @title Scatter plot: Temperature vs Energy
plt.figure(figsize=(7,4))

plt.scatter(
    df['temp'],
    df['energy'],
    color='tab:blue',
    alpha=0.5,
    edgecolor='k'
)

plt.xlabel('Temperature (Celsius)')
plt.ylabel('Daily energy (kWh)')
plt.title('Scatter Plot: Temperature vs Energy')

plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# @title
# --- Correlation comparison ---

from scipy.stats import pearsonr, spearmanr, kendalltau
import numpy as np
from xicorpy import compute_xi_correlation

x = df['temp'].values
y = df['energy'].values

# Clean data (important)
mask = ~np.isnan(x) & ~np.isnan(y)
x = x[mask]
y = y[mask]

# 1. Pearson
pearson_corr, _ = pearsonr(x, y)

# 2. Spearman
spearman_corr, _ = spearmanr(x, y)

# 3. Kendall
kendall_corr, _ = kendalltau(x, y)

# 4. Xi correlation
xi_corr = compute_xi_correlation(x, y)
xi_corr = xi_corr.item()

# Print results
print("=== Correlation Comparison ===")
print(f"Pearson   : {pearson_corr:.3f}")
print(f"Spearman  : {spearman_corr:.3f}")
print(f"Kendall   : {kendall_corr:.3f}")
print(f"Xi (xi)   : {xi_corr:.3f}")
```
